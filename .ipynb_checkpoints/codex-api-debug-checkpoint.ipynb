# Codex API 调用调试手册

通过分析 `~/.codex/logs_2.sqlite` 日志数据库，查看 Codex 桌面 App 的后端 API 调用细节。

> 💡 按 **Shift + Enter** 逐步运行每个 Cell

```python
# 🔧 初始化 - 连接数据库
import sqlite3
import json
import re
from datetime import datetime

DB_PATH = "/Users/weike/.codex/logs_2.sqlite"
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

# 查看数据库基本信息
cursor.execute("SELECT COUNT(*) FROM logs")
total = cursor.fetchone()[0]
print(f"✅ 数据库连接成功")
print(f"📊 总日志数: {total:,} 条")

# 查看最新日志时间
cursor.execute("SELECT ts FROM logs ORDER BY id DESC LIMIT 1")
latest_ts = cursor.fetchone()[0]
print(f"⏰ 最新日志时间: {datetime.fromtimestamp(latest_ts/1000)}")
```

## 1. 查看日志模块分布

先了解有哪些模块在打日志，找到 HTTP 相关的模块。

```python
# 📋 所有日志模块及其数量
cursor.execute("""
    SELECT target, COUNT(*) as cnt 
    FROM logs 
    GROUP BY target 
    ORDER BY cnt DESC
""")
rows = cursor.fetchall()
print(f"共 {len(rows)} 个日志模块:\n")
for target, cnt in rows:
    print(f"  {target:50s} {cnt:>6,}")
```

## 2. 查看最近的请求 / 响应

查找 HTTP 请求相关日志，查看最近的 API 调用。

```python
# 🔍 查找 HTTP 请求日志
cursor.execute("""
    SELECT id, ts, target, level, message
    FROM logs
    WHERE target LIKE '%http%' 
       OR target LIKE '%request%'
       OR target LIKE '%client%'
       OR message LIKE '%POST%'
       OR message LIKE '%GET%'
    ORDER BY id DESC
    LIMIT 20
""")
rows = cursor.fetchall()
print(f"找到 {len(rows)} 条 HTTP 相关日志:\n")
for log_id, ts, target, level, msg in rows:
    time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
    print(f"【{log_id}】 {time_str}  [{level}] {target}")
    print(f"    {msg[:200]}")
    print()
```

## 3. 分析请求体与模型信息

从日志中提取 API 请求的模型、消息数、工具等信息。

```python
# 📤 分析最近的请求详情
cursor.execute("""
    SELECT id, ts, target, message
    FROM logs
    WHERE message LIKE '%/chat/completions%'
       OR message LIKE '%POST%'
       OR target LIKE '%request%'
    ORDER BY id DESC
    LIMIT 10
""")
rows = cursor.fetchall()
print(f"最近 {len(rows)} 次请求:\n")
for i, (log_id, ts, target, msg) in enumerate(rows, 1):
    time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
    print(f"【{i}】 id={log_id}  {time_str}")

    # 尝试提取 URL
    url_match = re.search(r'(https?://[^\s]+)', msg)
    url = url_match.group(1) if url_match else 'N/A'
    print(f"     URL:   {url}")

    # 尝试提取模型名
    model_match = re.search(r'["\']model["\']\s*:\s*["\']([^"\']+)["\']', msg)
    model = model_match.group(1) if model_match else 'N/A'
    print(f"     模型:  {model}")

    # 尝试提取消息数
    msg_match = re.search(r'"messages"\s*:\s*\[', msg)
    msg_count = 'N/A'
    if msg_match:
        # 粗略统计消息数
        role_count = len(re.findall(r'"role"\s*:', msg))
        msg_count = str(role_count)
    print(f"     消息数: {msg_count}")
    print()
```

## 4. 查看完整的消息内容

提取最近一次对话的完整消息列表。

```python
# 💬 提取最近一次完整对话
cursor.execute("""
    SELECT id, ts, message
    FROM logs
    WHERE message LIKE '%"messages"%'
    ORDER BY id DESC
    LIMIT 5
""")
rows = cursor.fetchall()

for log_id, ts, msg in rows:
    # 尝试从日志中提取 JSON
    json_match = re.search(r'\{.*"messages".*\}', msg, re.DOTALL)
    if json_match:
        try:
            data = json.loads(json_match.group(0))
            messages = data.get('messages', [])
            time_str = datetime.fromtimestamp(ts/1000).strftime('%Y-%m-%d %H:%M:%S')
            print(f"📌 日志 ID: {log_id}  时间: {time_str}")
            print(f"   模型: {data.get("model", "N/A")}")
            print(f"总消息数: {len(messages)}\n")

            for j, m in enumerate(messages[-6:], 1):
                role = m.get('role', 'unknown')
                content = m.get('content', '')
                if isinstance(content, list):
                    content = ' '.join(
                        c.get('text', '') for c in content if isinstance(c, dict)
                    )
                content_preview = content[:300] + '...' if len(content) > 300 else content
                print(f"--- [{role}] ---")
                print(content_preview)
                print()
            break
        except json.JSONDecodeError:
            continue
else:
    print("未找到包含完整消息的日志")
```

## 5. 工具调用分析

查看请求中注册了哪些工具 / Function Calling。

```python
# 🛠️  工具调用分析
cursor.execute("""
    SELECT id, ts, message
    FROM logs
    WHERE message LIKE '%"tools"%'
       OR message LIKE '%function_call%'
       OR message LIKE '%tool_calls%'
    ORDER BY id DESC
    LIMIT 5
""")
rows = cursor.fetchall()

for log_id, ts, msg in rows:
    json_match = re.search(r'\{.*"tools".*\}', msg, re.DOTALL)
    if json_match:
        try:
            data = json.loads(json_match.group(0))
            tools = data.get('tools', [])
            time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
            print(f"📌 日志 ID: {log_id}  时间: {time_str}")
            print(f"已注册工具数: {len(tools)}\n")

            for t in tools:
                func = t.get('function', {})
                name = func.get('name', 'unknown')
                desc = func.get('description', '')[:100]
                print(f"  • {name}")
                print(f"    {desc}")
    print()
    break
else:
    print("未找到工具调用日志")
```

## 6. 响应状态与延迟

查看 API 响应的状态码和耗时。

```python
# ⏱️  响应状态与延迟分析
cursor.execute("""
    SELECT id, ts, target, level, message
    FROM logs
    WHERE target LIKE '%response%'
       OR target LIKE '%http%'
       OR message LIKE '%status%'
       OR message LIKE '%200%'
    ORDER BY id DESC
    LIMIT 20
""")
rows = cursor.fetchall()
print(f"最近 {len(rows)} 条响应相关日志:\n")

for log_id, ts, target, level, msg in rows:
    time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
    status_match = re.search(r'status[=:]\s*(\d+)', msg, re.IGNORECASE)
    if status_match:
        print(f"【{log_id}】 {time_str}  [{level}] {target}")
        print(f"     状态: {status_match.group(1)}")

    url_match = re.search(r'(https?://[^\s]+)', msg)
    if url_match:
        print(f"     URL:  {url_match.group(1)}")

    if status_match or url_match:
        print(f"     {msg[:150]}")
        print()
```

## 7. SSE 流式响应事件

如果使用了流式输出，查看事件类型分布。

```python
# 🌊 SSE 事件分析
cursor.execute("""
    SELECT id, ts, message
    FROM logs
    WHERE message LIKE '%data: %'
       OR message LIKE 'event: %'
       OR target LIKE '%stream%'
    ORDER BY id DESC
    LIMIT 200
""")
rows = cursor.fetchall()

if rows:
    print(f"📌 响应起始 id: {rows[-1][0]}\n")

    event_types = {}
    for log_id, ts, msg in rows:
        ev_match = re.search(r'event:\s*(\w+)', msg)
        if ev_match:
            evt = ev_match.group(1)
            event_types[evt] = event_types.get(evt, 0) + 1

    events = list(event_types.items())
    print(f"事件总数（前200）: {len(events)}\n")

    if events:
        for t, c in sorted(events, key=lambda x: -x[1]):
            print(f"   {t:45s} {c:>4} 次")
    else:
        # 查看内容样本
        print('未解析到 event 类型，前 5 条样本：')
        for log_id, ts, msg in rows[:5]:
            print(f"   {msg[:120]}")
else:
    print("未找到流式事件日志")
```

## 8. 完整响应输出

查看最近一次完整的响应内容（含 tool_calls / content）。

```python
# 📦 完整响应体分析
cursor.execute("""
    SELECT id, ts, message
    FROM logs
    WHERE message LIKE '%choices%'
       OR message LIKE '%assistant%'
    ORDER BY id DESC
    LIMIT 10
""")
rows = cursor.fetchall()

for log_id, ts, msg in rows:
    # 尝试提取 JSON
    for pattern in [r'\{.*"choices".*\}', r'\{.*"content".*\}']:
        m = re.search(pattern, msg, re.DOTALL)
        if m:
            try:
                data = json.loads(m.group(0))
                choices = data.get('choices', [])
                if not choices:
                    continue

                time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
                print(f"📌 日志 ID: {log_id}  时间: {time_str}")
                print(f"   模型: {data.get("model", "N/A")}")
                print(f"📦 输出项 ({len(choices)} 个):")

                for idx, choice in enumerate(choices):
                    msg_obj = choice.get('message', choice.get('delta', {}))
                    role = msg_obj.get('role', 'unknown')
                    content = msg_obj.get('content', '')
                    tool_calls = msg_obj.get('tool_calls', [])

                    print(f"\n--- choice[{idx}] role={role} ---")
                    if content:
                        print(content[:500])
                    if tool_calls:
                        print(f"\n🛠️  工具调用 ({len(tool_calls)} 个):")
                        for item in tool_calls:
                            print(f"       name: {item.get("name")}")
                            args = item.get('arguments', '')
                            if isinstance(args, str):
                                print(f"       args: {args[:200]}")
                            else:
                                print(f"       args: {json.dumps(args)[:200]}")
                break
            except json.JSONDecodeError:
                continue
    else:
        continue
    break
else:
    print("未找到完整响应日志")
```

## 9. 自定义搜索

根据关键词搜索日志内容。

```python
# 🔎 自定义关键词搜索
keyword = "model"  # 👈 修改这里的关键词

cursor.execute("""
    SELECT id, ts, target, level, message
    FROM logs
    WHERE message LIKE ?
    ORDER BY id DESC
    LIMIT 20
""", (f"%{keyword}%",))
rows = cursor.fetchall()
print(f"找到 {len(rows)} 条匹配日志:\n")

for log_id, ts, target, level, msg in rows:
    time_str = datetime.fromtimestamp(ts/1000).strftime('%H:%M:%S')
    print(f"【{log_id}】 {time_str}  [{level}] {target}")
    print(f"   {msg[:200]}")
    print()
```
